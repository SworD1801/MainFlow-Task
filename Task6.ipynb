```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Read the CSV file
file_path = 'C:\\Users\\Shubham Raj\\Desktop\\Data Analysis\\Data Set\\disney_plus_titles.csv'
data = pd.read_csv(file_path)
```

```python
# Display the first few rows of the dataframe
print(data.head())
```

```python
# 1. Basic Overview and Descriptive Statistics

# Distribution of content types
content_type_distribution = data['type'].value_counts()
print("Distribution of content types:\n", content_type_distribution)
```

```python
# Plot the distribution of content types
plt.figure(figsize=(8, 6))
sns.countplot(data['type'])
plt.title('Distribution of Content Types')
plt.xlabel('Type')
plt.ylabel('Count')
plt.show()
```

```python
# Most common genres
most_common_genres = data['listed_in'].value_counts().head(10)
print("Most common genres:\n", most_common_genres)
```

```python
# Plot the most common genres
plt.figure(figsize=(10, 6))
most_common_genres.plot(kind='bar')
plt.title('Most Common Genres')
plt.xlabel('Genre')
plt.ylabel('Count')
plt.show()
```

```python
# Distribution of release years
release_year_distribution = data['release_year'].value_counts().sort_index()
print("Distribution of release years:\n", release_year_distribution)
```

```python
# Plot the distribution of release years
plt.figure(figsize=(12, 6))
release_year_distribution.plot(kind='bar')
plt.title('Distribution of Release Years')
plt.xlabel('Release Year')
plt.ylabel('Count')
plt.show()
```

```python
# 2. Content Analysis

# Most frequent directors
most_frequent_directors = data['director'].value_counts().head(10)
print("Most frequent directors:\n", most_frequent_directors)
```

```python
# Plot the most frequent directors
plt.figure(figsize=(10, 6))
most_frequent_directors.plot(kind='bar')
plt.title('Most Frequent Directors')
plt.xlabel('Director')
plt.ylabel('Count')
plt.show()
```

```python
# Most frequent actors/actresses
most_frequent_actors = data['cast'].str.split(', ').explode().value_counts().head(10)
print("Most frequent actors/actresses:\n", most_frequent_actors)
```

```python
# Plot the most frequent actors/actresses
plt.figure(figsize=(10, 6))
most_frequent_actors.plot(kind='bar')
plt.title('Most Frequent Actors/Actresses')
plt.xlabel('Actor/Actress')
plt.ylabel('Count')
plt.show()
```

```python
# Most common ratings
most_common_ratings = data['rating'].value_counts()
print("Most common ratings:\n", most_common_ratings)
```

```python
# Plot the most common ratings
plt.figure(figsize=(8, 6))
sns.countplot(data['rating'])
plt.title('Most Common Ratings')
plt.xlabel('Rating')
plt.ylabel('Count')
plt.show()
```

```python
# 3. Time-Based Analysis

# Number of new releases over the years
new_releases_per_year = data['release_year'].value_counts().sort_index()
print("Number of new releases per year:\n", new_releases_per_year)
```

```python
# Plot the number of new releases over the years
plt.figure(figsize=(12, 6))
new_releases_per_year.plot(kind='line')
plt.title('Number of New Releases Over the Years')
plt.xlabel('Year')
plt.ylabel('Number of Releases')
plt.show()
```

```python
# Seasonal trends in release dates
data['date_added'] = pd.to_datetime(data['date_added'], errors='coerce')
data['month'] = data['date_added'].dt.month
seasonal_trends = data['month'].value_counts().sort_index()
print("Seasonal trends in release dates:\n", seasonal_trends)
```

```python
# Plot the seasonal trends in release dates
plt.figure(figsize=(10, 6))
seasonal_trends.plot(kind='bar')
plt.title('Seasonal Trends in Release Dates')
plt.xlabel('Month')
plt.ylabel('Number of Releases')
plt.show()
```

```python
# 4. Content Duration

# Distribution of content duration
duration_distribution = data['duration'].dropna().str.extract('(\d+)').astype(int)
print("Distribution of content duration:\n", duration_distribution.describe())
```

```python
# Plot the distribution of content duration
plt.figure(figsize=(10, 6))
sns.histplot(duration_distribution[0], bins=20)
plt.title('Distribution of Content Duration')
plt.xlabel('Duration (minutes)')
plt.ylabel('Count')
plt.show()
```

```python
# Differences in duration between movies and TV shows
movies_duration = data[data['type'] == 'Movie']['duration'].dropna().str.extract('(\d+)').astype(int)
tv_shows_duration = data[data['type'] == 'TV Show']['duration'].dropna().str.extract('(\d+)').astype(int)

print("Movies duration statistics:\n", movies_duration.describe())
print("TV shows duration statistics:\n", tv_shows_duration.describe())
```

```python
# Plot the duration comparison
plt.figure(figsize=(12, 6))
sns.histplot(movies_duration[0], color='blue', label='Movies', kde=True, stat="density", common_norm=False)
sns.histplot(tv_shows_duration[0], color='green', label='TV Shows', kde=True, stat="density", common_norm=False)
plt.title('Duration Comparison between Movies and TV Shows')
plt.xlabel('Duration (minutes)')
plt.ylabel('Density')
plt.legend()
plt.show()
```
